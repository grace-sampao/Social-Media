```python
import environment
import importlib

importlib.reload(environment)

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_tavily import TavilyExtract
from langchain.agents import create_agent
```

```python
# Initialize the llm
llm = ChatGoogleGenerativeAI(
  model="gemini-3.5-flash",
  api_key=environment.GEMINI_API_KEY,
  temperature=1.0,
  max_tokens=None,
  timeout=None,
  max_retries=2,
)

# Set up the web extraction tool
tavily_extract_tool = TavilyExtract(
  extract_depth="basic",
  include_images=False,
  tavily_api_key=environment.TAVILY_API_KEY,
  # format="text",
)
```

```python
system_prompt_technical_blog_posts = """
You are a skilled LinkedIn social media manager.
You write a LinkedIn post, that appears in relevant feeds and search results, 
explaining and summarizing a blog post you have extracted from my technical blog 
website.

The LinkedIn post you write is optimized for the LinkedIn platform, uses appropriate emojis, 
includes a link to the extracted blog post and is written in first person as if 
I was the one who wrote it. Remember to use relevant hashtags.
"""

url = "['https://grace-sampao.github.io/posts/building-llm-powered-applications-chapter-02/']"
```

```python
system_prompt_linkedin_post = """
You are a skilled LinkedIn social media manager.

Each post you write MUST meet all the criteria below:
 - Highly optimized for the platform's algorithm.
 - Appears in relevant feeds and search results.
 - Drives engagement on the platform.
 - Uses relevant hashtags.
 - Uses appropriate and relevant emojis.
 - Written in first person format.

Using your LinkedIn social media management skills and strictly adhering to the 
above criteria, write a LinkedIn post from my personal notes that I will share 
with you.

Ensure to retain the sentiment and writing style depicted in my personal notes.

Personal Notes:
"""
```

```python
agent = create_agent(
  model=llm,
  system_prompt=system_prompt_linkedin_post
)
```

```python
# Invocation
# linkedin_post = agent.invoke(
#   {"messages": url}
# )
```

```python
linkedin_post = agent.invoke(
  {
    "messages": [
      {
        "role": "user",
        "content": personal_notes
			}
		]
	}
)
```
